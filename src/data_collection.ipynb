```python
# Importing the necessary Libraries
import os
import pandas as pd
from pathlib import Path
```

```python
# Loading the previous results and goalscorers 

base_dir = Path.cwd()
while base_dir.name != "World Cup Predictor":
    base_dir = base_dir.parent
results_path = base_dir/"data"/"raw"/"results.csv"
goalscorers_path = base_dir/"data"/"raw"/"goalscorers.csv"
prev_results_df = pd.read_csv(results_path)
goalscorers_df = pd.read_csv(goalscorers_path)
```

```python
# Checking if the data is loaded properly

print(prev_results_df.shape)
print(prev_results_df.head())
print(prev_results_df['tournament'].unique())
print(prev_results_df.isnull().sum())
print(prev_results_df['date'].min(), prev_results_df['date'].max())
```

#### Filtering the data 
##### We need to filter **prev_results_df** into two separate DataFrames. The World Cup only subset is the ground truth for simulating the 2026 tournament bracket, only WC matches matter for that.
##### The competetive matches (no friendlies) subset is for calculating team strength. Friendlies are often low-stakes with rotated squads, so they'd skew the model's goal averges and make the predictions less accurate.

```python
# World Cup matches dataframe
wc_matches_df = prev_results_df[prev_results_df["tournament"] == "FIFA World Cup"]
wc_matches_path = base_dir/"data"/"processed"/"wc_matches.csv"
wc_matches_df.to_csv(wc_matches_path, index = False)

# Non-friendly Competitive matches dataframe
competitive_matches_df = prev_results_df[(prev_results_df["tournament"] != "FIFA World Cup") & (prev_results_df["tournament"] != "Friendly")]
competitive_matches_path = base_dir/"data"/"processed"/"competitive_matches.csv"
competitive_matches_df.to_csv(competitive_matches_path, index = False)
```
