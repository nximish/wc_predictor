```python
# Importing the necessary Libraries
import os
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Loading the World Cup Matches and Competitive Matches 

base_dir = Path.cwd()
while base_dir.name != "World Cup Predictor":
    base_dir = base_dir.parent
wc_matches_path = base_dir/"data"/"processed"/"wc_matches.csv"
competitive_matches_path = base_dir/"data"/"processed"/"competitive_matches.csv"
wc_matches_df = pd.read_csv(wc_matches_path)
competitive_matches_df = pd.read_csv(competitive_matches_path)
```

```python
# Converting date to  DateTime format

wc_matches_df["date"] = pd.to_datetime(wc_matches_df["date"])
competitive_matches_df["date"] = pd.to_datetime(competitive_matches_df["date"])
```

```python
# Remove matches with no score (walkovers/unplayed fixtures)

wc_matches_df = wc_matches_df.dropna(subset=['home_score','away_score'])
competitive_matches_df = competitive_matches_df.dropna(subset=['home_score','away_score'])
```

```python
# Adding a "result" column to assign which team won or if the game was a draw

wc_conditions = [wc_matches_df['home_score'] > wc_matches_df['away_score'], wc_matches_df['home_score'] < wc_matches_df['away_score']]
wc_choices = ["H", "A"]
wc_matches_df['result'] = np.select(wc_conditions, wc_choices, default= "D")

competitive_conditions = [competitive_matches_df['home_score'] > competitive_matches_df['away_score'], competitive_matches_df['home_score'] < competitive_matches_df['away_score']]
competitive_choices = ["H", "A"]
competitive_matches_df['result'] = np.select(competitive_conditions, competitive_choices, default= "D")
```

```python
# Check if the addition of "result" worked
# print(wc_matches_df[['home_team', 'away_team', 'home_score', 'away_score', 'result']].head(10))
```

```python
# Adding a "total_goals" column

wc_matches_df['total_goals'] = wc_matches_df['home_score'] + wc_matches_df['away_score']
competitive_matches_df['total_goals'] = competitive_matches_df['home_score'] + competitive_matches_df['away_score']
```

```python
# Check if the addition of "total_goals" worked
# print(wc_matches_df[['home_team', 'away_team', 'home_score', 'away_score', 'result', 'total_goals']].head(10))
```

```python
# Saving this cleaned dataframes to the same "processed folder"
clean_wc_matches_path = base_dir/"data"/"processed"/"clean_wc_matches.csv"
wc_matches_df.to_csv(clean_wc_matches_path, index = False)

clean_competitive_matches_path = base_dir/"data"/"processed"/"clean_competitive_matches.csv"
competitive_matches_df.to_csv(clean_competitive_matches_path, index = False)
```
