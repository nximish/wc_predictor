```python
# Importing the necessary Libraries
import os
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
elo_df = pd.read_csv('https://www.eloratings.net/World.tsv', sep='\t', header=None)
elo_df = elo_df[[2,3]]
elo_df = elo_df.rename(columns={
    2: 'country_code',
    3: 'elo_rating'
})
```

```python
teams_df = pd.read_csv('https://www.eloratings.net/en.teams.tsv', on_bad_lines='skip', sep='\t', header=None)
teams_df = teams_df.rename(columns={
    0: 'country_code',
    1: 'name'
})
```

```python
elo_ratings_df = pd.merge(elo_df, teams_df, on='country_code')
```

```python
elo_ratings_df['name'] = elo_ratings_df['name'].replace({'Czechia': 'Czech Republic'})
```

```python
base_dir = Path.cwd()
while base_dir.name != "World Cup Predictor":
    base_dir = base_dir.parent

# Save the Elo Ratings dataframe to the raw data folder

elo_ratings_path = base_dir/"data"/"raw"/"elo_ratings.csv"
elo_ratings_df.to_csv(elo_ratings_path, index = True)
```
