```python
import requests
import pandas as pd

API_KEY = "4e3cb7a456264e27a695093188ed48e3"

headers = {
  "X-Auth-Token": API_KEY
}

url = "https://api.football-data.org/v4/competitions/WC/matches"

response = requests.get(url, headers=headers)
data = response.json()

matches = []
for match in data['matches']:
  matches.append({
      'home': match['homeTeam']['name'],
      'away': match['awayTeam']['name'],
      'date': match['utcDate'],
      'stage': match['stage'],
      'group': match.get('group', None)
  })

df_fixtures_2026 = pd.DataFrame(matches)
print(df_fixtures_2026.shape)
print(df_fixtures_2026.head())
```

```python
from pathlib import Path

base_dir = Path.cwd()
while base_dir.name != "World Cup Predictor":
    base_dir = base_dir.parent


df_fixtures_2026.to_csv(base_dir/"data"/"raw"/"fixtures_2026.csv", index=False)

# For processed, clean the date column and split group stage from knockout fixtures:

df_fixtures_2026['date'] = pd.to_datetime(df_fixtures_2026['date'])

df_group_stage = df_fixtures_2026[df_fixtures_2026['stage'] == 'GROUP_STAGE']
df_knockouts = df_fixtures_2026[df_fixtures_2026['stage'] != 'GROUP_STAGE']

df_group_stage.to_csv(base_dir/"data"/"processed"/"fixtures_group_stage.csv", index=False)
df_knockouts.to_csv(base_dir/"data"/"processed"/"fixtures_knockouts.csv", index=False)
```
