```python
# Importing the necessary Libraries
import os
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
base_dir = Path.cwd()
while base_dir.name != "World Cup Predictor":
    base_dir = base_dir.parent
```

```python
dfs = []
for page in range(1, 10):
  url = f'https://www.transfermarkt.co.uk/statistik/weltrangliste?page={page}'
  dfs.append(pd.read_html(url)[1])
df = pd.concat(dfs, ignore_index=True)
```

```python
df = df.rename(columns={'#': 'fifa_ranking', 'Nation': 'name', 'Total value': 'total_market_value'})
```

```python
df.to_csv(base_dir/'data'/'raw'/'national_teams.csv', index=False)
```
